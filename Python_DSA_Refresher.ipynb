```python
days = int(input('Input number of days: ')) #Get input for number of days
seconds = days*24*60*60 #do the math to convert the number of days into seconds
print(f"Number of seconds: ", {seconds}) # output the number of seconds
```

```python
radius = int(input("Input the radius of the sphere in centimeters: "))
volume = (4/3) * (22/7) * (radius**3)
print(f"The volume of the sphere is: {volume} cubic centimeters")
```

```python
def compute():
  length = int(input("What is the length one side of the square in centimeters: "))
  perimeter = length * 4
  area = length ** 2

  print(f"Perimeter: {perimeter}")
  print(f"Area: {area}")

compute()
```

```python
def upper_or_lower(character):
  if character.isupper():
    print("Character is uppercase")
  else:
    print("Character is lowercase")

upper_or_lower("C")
```
